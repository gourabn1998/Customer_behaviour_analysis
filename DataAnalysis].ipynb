```python
group_obj = df.groupby('Category')
for category, data in group_obj:
    print ('category :', category)
    print ('data :', data)
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform (lambda x: x.fillna(x.mean()))
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df['age_category'] = pd.qcut (df['age'], q=4, labels = ['young','adults', 'middle_aged','senior'])
```
